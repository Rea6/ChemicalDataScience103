# Numpy recap: Quick demo

```python
import numpy as np
```

Create a numpy array:

```python
twodim = np.random.random((30, 5))
onedim = np.arange(1,20,1)
```

Explore attributes:

```python
twodim.ndim
twodim.size
twodim.shape
```

Access elements:

```python
twodim[(1,1)]
```

Slicing:

```python
onedim[::-1]
```

```python
twodim[::3,2::]
```

Ufuncs / aggregates

```python
%timeit multiplied = np.multiply(onedim, 2)
```

```python
%timeit multiplied = onedim*2
```

```python
np.sum(twodim, axis=0)
```

```python
%timeit sum(twodim)
%timeit np.sum(twodim)
```

```python
onedim
```

```python
onedim < 10
```
